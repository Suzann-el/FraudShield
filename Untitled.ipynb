# 3 FeatureEngineering






###  Encodage des catégorielles

```python
cat_cols = ['ProductCD', 'card4', 'card6', 'P_emaildomain',
'R_emaildomain', 'DeviceType', 'DeviceInfo']
for col in cat_cols:
  df[col] = df[col].astype('category' )
```

### Features temporelles

```python
df['heure_tx'] = (df['TransactionDT'] // 3600) % 24
df['jour_semaine'] = (df['TransactionDT'] // (3600 * 24)) % 7
df['est_nuit'] = ((df['heure_tx'] >= 22) | (df['heure_tx'] <=5)).astype(int)
```

### Velocity check (fréquence par email) 

```python

df = df.sort_values('TransactionDT')
df['tx_rank_email'] = df.groupby('P_emaildomain').cumcount()
```

### Agrégation par carte (comportement historique)


```python
agg = df.groupby('card1')['TransactionAmt'].agg(montant_moyen_carte='mean',montant_std_carte='std').reset_index()
df = df.merge(agg, on='card1', how='left')
df['z_score_montant'] = ((df['TransactionAmt'] - df['montant_moyen_carte'])/ (df['montant_std_carte'] + 1e-6)) # montant anormalement élevé par rapport à l'historique de la carte
```

### Suppression features à forte cardinalité ou NaN > 90% 


```python
cols_to_drop = [
    c for c in df.columns
         if df[c].isnull().mean() > 0.90
         or c in ['TransactionID', 'TransactionDT']]
df.drop(columns=cols_to_drop, inplace=True)
print(f"Features retenues : {df.shape[1] - 1}")
```

```python
# Définition de X et y
X = df.drop(columns=['isFraud'])
y = df['isFraud']

print(f"X shape : {X.shape}")
print(f"Taux de fraude : {y.mean()*100:.2f}%")
```

### 4 Modélisation — LightGBM avec validation croisée


```python
import pandas as pd
import numpy as np
from sklearn.preprocessing import LabelEncoder
from sklearn.model_selection import StratifiedKFold
from sklearn.metrics import roc_auc_score, average_precision_score
import lightgbm as lgb

# ── 1. Encoder les colonnes object ──────────────────────────
cat_cols = [col for col in X.columns if X[col].dtype == 'object']
le = LabelEncoder()
for col in cat_cols:
    X[col] = X[col].fillna('missing')
    X[col] = le.fit_transform(X[col].astype(str))




```

```python
# ── 2. Paramètres LightGBM ──────────────────────────────────
params = {
    'objective'        : 'binary',
    'metric'           : 'auc',
    'scale_pos_weight' : (y == 0).sum() / (y == 1).sum(),
    'learning_rate'    : 0.05,
    'num_leaves'       : 63,
    'min_child_samples': 100,
    'feature_fraction' : 0.8,
    'bagging_fraction' : 0.8,
    'bagging_freq'     : 5,
    'verbose'          : -1
}
```

```python
# ── 3. Validation croisée stratifiée (5 folds) ──────────────
skf       = StratifiedKFold(n_splits=5, shuffle=True, random_state=42)
oof_preds = np.zeros(len(X))

for fold, (train_idx, val_idx) in enumerate(skf.split(X, y)):
    X_tr, X_val = X.iloc[train_idx], X.iloc[val_idx]
    y_tr, y_val = y.iloc[train_idx], y.iloc[val_idx]

    dtrain = lgb.Dataset(X_tr, label=y_tr)
    dval   = lgb.Dataset(X_val, label=y_val, reference=dtrain)

    model = lgb.train(
        params, dtrain,
        num_boost_round=1000,
        valid_sets=[dval],
        callbacks=[lgb.early_stopping(50), lgb.log_evaluation(100)]
    )

    oof_preds[val_idx] = model.predict(X_val)
    fold_auc = roc_auc_score(y_val, oof_preds[val_idx])
    print(f"Fold {fold+1} — AUC = {fold_auc:.4f}")
```

```python
# ── 4. Métriques globales ────────────────────────────────────
print(f"\nAUC OOF global : {roc_auc_score(y, oof_preds):.4f}")
print(f"PR-AUC         : {average_precision_score(y, oof_preds):.4f}")
```

```python
# Échantillon de 5000 lignes — suffisant pour le summary plot
X_sample = X_val.sample(5000, random_state=42)

explainer   = shap.TreeExplainer(model)
shap_values = explainer.shap_values(X_sample)

shap.summary_plot(shap_values, X_sample, plot_type='bar', max_display=20)
```

## 5-Explicabilité SHAP




```python
# Summary plot — importance globale
shap.summary_plot(shap_values, X_val, plot_type='bar',
max_display=20, show=False)
plt.tight_layout()
plt.savefig('shap_importance.png', dpi=150)
plt.show()
```

```python
# Force plot — expliquer une fraude spécifique
idx_fraude = np.where(y_val == 1)[0][0]
shap.force_plot(
explainer.expected_value,
    shap_values[idx_fraude],
    X_val.iloc[idx_fraude],
    matplotlib=True
)
plt.savefig('shap_force_plot.png', dpi=150, bbox_inches='tight')
plt.show()
```
